## Conjuction

```python
print(prob(banker))

print(prob(democrat))

print(prob(banker & democrat))

# Demonstrating that it is commutative

print(prob(democrat & banker))
```

## Conditional Probability

```python
# Think of this like P(d | l)

selected = democrat[liberal]

prob(selected)
```

```python
# P(b | d)

selected = banker[democrat]

prob(selected)
```

```python
# P(f | b)

selected = female[banker]

prob(selected)
```

```python
def conditional(proposition, given):
    return prob(proposition[given])
```

```python
# P(l | f)

conditional(liberal, given = female)
```

## Conditional Probability is not Commutative

```python
conditional(banker, given = female)
```

## Condition and Conjunction

```python
conditional(liberal & female, given = banker)
```

## Laws of Probability

#### Theorem 1: Using a conjunction to compute a conditional probability.
#### Theorem 2: Using a conditional probability to compute a conjunction.
#### Theorem 3: Using conditional(A, B) to compute conditional(B, A).

```python
# Theorem 1

print(female[banker].mean())

print(conditional(female, given = banker))

print(prob(female & banker) / prob(banker))
```

#### Theorem 1

$$P(A \mid B) = \frac{P(A \text{ and } B)}{P(B)}$$

```python
# Theorem 2

print(prob(liberal & democrat))

print(prob(democrat) * conditional(liberal, democrat))
```

#### Theorem 2

$$P(A \text { and }  B) = P(B)\,P(A \mid B)$$

```python
# Theorem 3

print(conditional(liberal, banker))

print(prob(liberal) * conditional(banker, liberal) / prob(banker))
```

#### Theorem 3

$$ P(A \text{ and } B) = P(B \text{ and } A) $$

Apply theorum 2:

$$ P(B)\,P(A \mid B) = P(A)\, P(B \mid A) $$

Divide through by P(B)

$$ P(A \mid B) = \frac{P(A)\, P(B\mid A)}{P(B)} $$

That is Bayes's theorem. 

## Law of Total Probability

$$ P(A) = P(B_1 \text{ and } A) + P(B_2 \text{ and } A) $$

```python
male = (gss['sex'] == 1)

print(prob(banker))

print(prob(male & banker) + prob(female & banker))
```

Apply theorem 2 to this law:

$$ P(A) = P(B_1)\,P(A\mid B_1)+P(B_2)\,P(A \mid B_2) $$

```python
print(prob(male)*conditional(banker, male) + 
      prob(female)*conditional(banker, female))
```

#### Use this more concise format when there are more than two conditions:

$$ P(A) = \sum_i \, P(B_i) \, P(A \mid B_i) $$

```python
# Example

B = gss['polviews']
B.value_counts().sort_index()
```

```python
i = 4

prob(B==i) * conditional(banker, B==i)
```
