## 1.1 Real Numbers: Algebra Essentials

### Number sets

Each set contains the one before it:

$$
\mathbb{N} \subset \mathbb{W} \subset \mathbb{Z} \subset \mathbb{Q} \subset \mathbb{R}
$$

| Set | Symbol | Contents |
|---|---|---|
| Natural | $\mathbb{N}$ | $\{1, 2, 3, \dots\}$ |
| Whole | $\mathbb{W}$ | $\{0, 1, 2, 3, \dots\}$ |
| Integers | $\mathbb{Z}$ (book uses $I$) | $\{\dots, -2, -1, 0, 1, 2, \dots\}$ |
| Rational | $\mathbb{Q}$ | $\left\{ \frac{m}{n} : m, n \in \mathbb{Z},\ n \neq 0 \right\}$ |
| Irrational | $\mathbb{Q}'$ | real numbers that are not rational ($\pi$, $\sqrt{2}$, $e$) |

- A number is **rational if and only if** its decimal expansion terminates ($0.25$) or repeats ($0.\overline{3}$).
- Every real number is exactly one of rational or irrational.
- Simplify before classifying: $\sqrt{36} = 6$ is a natural number; $\sqrt{73}$ is irrational.

```python
fig, ax = plt.subplots(figsize=(8, 4.6))
ax.set_xlim(0, 10)
ax.set_ylim(0, 6)
ax.axis("off")

def box(x, y, w, h, color, label):
    ax.add_patch(FancyBboxPatch(
        (x, y), w, h, boxstyle="round,pad=0.02,rounding_size=0.25",
        facecolor=color + "18", edgecolor=color, linewidth=1.6))
    ax.text(x + 0.18, y + h - 0.2, label, color=color,
            fontsize=12, fontweight="bold", va="top")

box(0.1, 0.1, 9.8, 5.8, INK, r"$\mathbb{R}$  real")
box(0.4, 0.4, 6.0, 4.9, BLUE, r"$\mathbb{Q}$  rational")
box(0.7, 0.7, 4.0, 3.8, GREEN, r"$\mathbb{Z}$  integers")
box(1.0, 1.0, 2.7, 2.7, PURPLE, r"$\mathbb{W}$  whole")
box(1.3, 1.3, 1.5, 1.7, ORANGE, r"$\mathbb{N}$")
box(6.7, 0.4, 2.9, 4.9, GRAY, r"$\mathbb{Q}'$  irrational")

# examples in each ring
ax.text(2.05, 1.85, r"$1,\ 7$" + "\n" + r"$\sqrt{36}$",
        ha="center", va="center", color=ORANGE, fontsize=11)
ax.text(3.25, 1.6, r"$0$", ha="center", color=PURPLE, fontsize=12)
ax.text(4.2, 1.6, r"$-3$", ha="center", color=GREEN, fontsize=12)
ax.text(5.5, 2.6, r"$\frac{1}{2}$" + "\n\n" + r"$0.\overline{3}$" + "\n\n" + r"$-0.25$",
        ha="center", va="center", color=BLUE, fontsize=12)
ax.text(8.15, 2.6, r"$\pi$" + "\n\n" + r"$\sqrt{2}$" + "\n\n" + r"$\sqrt{73}$",
        ha="center", va="center", color=GRAY, fontsize=12)
plt.show()
```

*Each set sits inside the next. The irrationals are a separate region of $\mathbb{R}$, not a ring around $\mathbb{Q}$.*

### Order of operations

1. **Grouping:** parentheses, brackets, *and* fraction bars, radicals, absolute value bars.
2. **Exponents and radicals.**
3. **Multiplication and division**, left to right.
4. **Addition and subtraction**, left to right.

### Properties of real numbers

For all real $a, b, c$:

| Property | Addition | Multiplication |
|---|---|---|
| Commutative | $a + b = b + a$ | $ab = ba$ |
| Associative | $(a + b) + c = a + (b + c)$ | $(ab)c = a(bc)$ |
| Identity | $a + 0 = a$ | $a \cdot 1 = a$ |
| Inverse | $a + (-a) = 0$ | $a \cdot \frac{1}{a} = 1$, for $a \neq 0$ |
| Distributive | $a(b + c) = ab + ac$ | |

Subtraction and division are **neither** commutative nor associative. Subtracting a sum flips the sign of every term: $a - (b + c) = a - b - c$.

**Worked example.**

$$
2r - 5(3 - r) + 4 = 2r - 15 + 5r + 4 = 7r - 11
$$

:::{admonition} Common mistakes
:class: warning
- $-3^2 = -9$, but $(-3)^2 = 9$. The exponent applies only to what it touches.
- $12 \div 3 \cdot 2 = 8$, not $2$. Multiplication and division go left to right, with equal priority.
- $5 - (x - 2) = 7 - x$. The minus distributes to *both* terms.
:::

:::{tip} Where this shows up
These properties are what justify every algebra step you take. In real analysis (Abbott), $\mathbb{R}$ is built from essentially this table as axioms.
:::

## 1.2 Exponents and Scientific Notation

### Rules of exponents

For nonzero $a, b$ and integers $m, n$:

| Rule | Statement |
|---|---|
| Product | $a^m \cdot a^n = a^{m+n}$ |
| Quotient | $\dfrac{a^m}{a^n} = a^{m-n}$ |
| Power | $(a^m)^n = a^{mn}$ |
| Zero | $a^0 = 1$ |
| Negative | $a^{-n} = \dfrac{1}{a^n}$ |
| Power of a product | $(ab)^n = a^n b^n$ |
| Power of a quotient | $\left(\dfrac{a}{b}\right)^n = \dfrac{a^n}{b^n}$ |

- **Product rule adds** exponents (same base, multiplied). **Power rule multiplies** exponents (a power raised to a power).
- A factor with a negative exponent moves across the fraction bar and becomes positive: $\dfrac{x^{-3}}{y^{2}} = \dfrac{1}{x^3 y^2}$.
- $0^0$ is treated as undefined here.

```python
fig, ax = plt.subplots(figsize=(7.5, 4.2))
x = np.linspace(-3.4, 3.4, 400)
ax.plot(x, 2.0 ** x, color=BLUE, alpha=0.25, lw=2)
n = np.arange(-3, 4)
y = 2.0 ** n
ax.plot(n, y, "o", color=BLUE, ms=7, zorder=3)
labels = [r"$\frac{1}{8}$", r"$\frac{1}{4}$", r"$\frac{1}{2}$",
          "$1$", "$2$", "$4$", "$8$"]
for xi, yi, lab in zip(n, y, labels):
    ax.annotate(lab, (xi, yi), textcoords="offset points",
                xytext=(0, 10), ha="center", fontsize=12, color=INK)
ax.plot(0, 1, "o", color=ORANGE, ms=9, zorder=4)
ax.annotate(r"$2^0 = 1$", (0, 1), xytext=(-1.9, 4.2), fontsize=11,
            color=ORANGE, arrowprops=dict(arrowstyle="->", color=ORANGE))
ax.annotate("", xy=(-1.5, 0.5), xytext=(0.5, 1.8),
            arrowprops=dict(arrowstyle="->", color=GRAY,
                            connectionstyle="arc3,rad=0.3"))
ax.text(-2.9, 2.0, r"each step left: $\div 2$", color=GRAY, fontsize=10)
ax.set_xticks(n)
ax.set_xticklabels([f"${k}$" for k in n])
ax.set_xlabel(r"exponent $n$")
ax.set_ylabel(r"$2^n$")
ax.set_ylim(0, 9.5)
plt.show()
```

*Stepping the exponent down by 1 divides by 2 every time. Keep going past $2^1$ and you land on $2^0 = 1$, then $2^{-1} = \tfrac{1}{2}$, $2^{-2} = \tfrac{1}{4}$. That pattern is *why* $a^0 = 1$ and $a^{-n} = \dfrac{1}{a^n}$.*

**Worked example.**

$$
\left(\frac{2x^{3}}{y^{-2}}\right)^{2} = \left(2x^3y^2\right)^2 = 4x^{6}y^{4}
$$

### Scientific notation

A number is in scientific notation when written as $a \times 10^n$ with $1 \le |a| < 10$ and $n \in \mathbb{Z}$.

- Large numbers get positive $n$; numbers less than 1 get negative $n$.
- When multiplying, multiply the $a$ parts and add the exponents, then renormalize: $(7 \times 10^4)(5 \times 10^6) = 35 \times 10^{10} = 3.5 \times 10^{11}$.

:::{admonition} Common mistakes
:class: warning
- $(a + b)^2 \neq a^2 + b^2$. Exponents distribute over **products**, never over sums.
- $2x^{-1} = \dfrac{2}{x}$, not $\dfrac{1}{2x}$. The exponent applies only to $x$.
- $x^2 \cdot x^3 = x^5$ (add), but $(x^2)^3 = x^6$ (multiply).
- $x^2 + x^3$ cannot be combined at all.
:::

:::{tip} Where this shows up
A likelihood is a product of many terms like $\theta^{y_i}(1 - \theta)^{1 - y_i}$, and collapsing it relies on the product rule. Exponent rules also underpin logarithms (Chapter 6) and the power rule for derivatives.
:::

## 1.3 Radicals and Rational Exponents

### Square roots

- The **principal square root** $\sqrt{a}$ (for $a \ge 0$) is the *nonnegative* number whose square is $a$. So $\sqrt{25} = 5$, not $\pm 5$.
- For any real $x$: $\sqrt{x^2} = |x|$.
- **Product rule:** $\sqrt{ab} = \sqrt{a}\,\sqrt{b}$ for $a, b \ge 0$.
- **Quotient rule:** $\sqrt{\dfrac{a}{b}} = \dfrac{\sqrt{a}}{\sqrt{b}}$ for $a \ge 0$, $b > 0$.
- **Simplifying** means pulling out perfect-square factors: $\sqrt{300} = \sqrt{100 \cdot 3} = 10\sqrt{3}$.
- **Adding:** only like radicals combine. $\sqrt{12} + 2\sqrt{3} = 2\sqrt{3} + 2\sqrt{3} = 4\sqrt{3}$.

```python
fig, (a1, a2) = plt.subplots(1, 2, figsize=(9.5, 3.9))
x = np.linspace(-3, 3, 400)

a1.plot(x, np.sqrt(x ** 2), color=BLUE, lw=2.4, label=r"$\sqrt{x^2} = |x|$")
a1.plot(x, x, color=ORANGE, lw=1.6, ls="--", label=r"$x$")
a1.fill_between(x[x < 0], x[x < 0], np.sqrt(x[x < 0] ** 2),
                color=ORANGE, alpha=0.08)
a1.text(-2.6, 0.9, "they disagree\nwhen $x < 0$", color=ORANGE, fontsize=10)
center_axes(a1)
a1.set_ylim(-3, 3.3)
a1.legend(loc="lower right", frameon=False)
a1.set_title(r"$\sqrt{x^2}$ is not $x$", color=INK)

xp = np.linspace(0, 4, 300)
xa = np.linspace(-4, 4, 400)
a2.plot(xp, np.sqrt(xp), color=BLUE, lw=2.4, label=r"$\sqrt{x}$  (even root)")
a2.plot(xa, np.cbrt(xa), color=GREEN, lw=2.0, label=r"$\sqrt[3]{x}$  (odd root)")
a2.plot(-1, -1, "o", color=GREEN, ms=6)
a2.annotate(r"$\sqrt[3]{-1} = -1$", (-1, -1), xytext=(-3.9, -0.55),
            color=GREEN, fontsize=10,
            arrowprops=dict(arrowstyle="->", color=GREEN))
a2.text(-3.9, 1.0, r"$\sqrt{x}$ undefined" + "\nfor $x < 0$", color=BLUE, fontsize=10)
center_axes(a2)
a2.set_ylim(-2, 2.4)
a2.legend(loc="lower right", frameon=False)
a2.set_title("Odd roots accept negatives", color=INK)
fig.tight_layout()
plt.show()
```

*Left: $\sqrt{x^2}$ is the V-shaped $|x|$, which matches $x$ only on the right half. Right: even roots exist only for $x \ge 0$; odd roots are defined everywhere.*

### Rationalizing denominators

- Single term: multiply by $\dfrac{\sqrt{b}}{\sqrt{b}}$.
- Two terms: multiply by the **conjugate**, which flips the sign of the radical term. This works because $(a + \sqrt{b})(a - \sqrt{b}) = a^2 - b$.

**Worked example.**

$$
\frac{1}{2 + \sqrt{3}} \cdot \frac{2 - \sqrt{3}}{2 - \sqrt{3}} = \frac{2 - \sqrt{3}}{4 - 3} = 2 - \sqrt{3}
$$

### $n$th roots and rational exponents

- $\sqrt[n]{a}$ is the number that, raised to the $n$th power, gives $a$.
- Odd roots of negatives are fine: $\sqrt[3]{-8} = -2$. Even roots of negatives are not real.
- Rational exponents:

$$
a^{1/n} = \sqrt[n]{a}, \qquad a^{m/n} = \left(\sqrt[n]{a}\right)^m = \sqrt[n]{a^m}
$$

  The numerator is the power, the denominator is the root. Taking the root first keeps numbers small: $8^{2/3} = (\sqrt[3]{8})^2 = 4$.
- All the exponent rules from §1.2 still apply.

:::{admonition} Common mistakes
:class: warning
- $\sqrt{a + b} \neq \sqrt{a} + \sqrt{b}$. Check: $\sqrt{9 + 16} = 5$, but $3 + 4 = 7$.
- $\sqrt{x^2} = |x|$, not $x$. This matters as soon as $x$ can be negative.
- $x^{-1/2} = \dfrac{1}{\sqrt{x}}$. A negative exponent means reciprocal, not a negative result.
:::

:::{tip} Where this shows up
Standard deviation is $\sqrt{\text{variance}}$, and the standard error of a mean is $\sigma / \sqrt{n}$. Multiplying by a conjugate is a standard trick for limits in Stewart.
:::

## 1.4 Polynomials

### Vocabulary

A **polynomial** has the form

$$
a_n x^n + a_{n-1} x^{n-1} + \dots + a_1 x + a_0,
$$

where every exponent is a **nonnegative integer**.

- The **degree** is the highest power. The **leading term** contains that power, and its coefficient is the **leading coefficient**.
- **Standard form** writes the powers in descending order.
- Monomial = 1 term; binomial = 2 terms; trinomial = 3 terms.
- $x^{-1}$ or $\sqrt{x}$ appearing anywhere means it is *not* a polynomial.

### Operations

- **Add or subtract:** combine like terms (same variables with the same exponents). When subtracting, distribute the minus across the entire second polynomial first.
- **Multiply:** every term in the first times every term in the second, then combine like terms. FOIL is just this rule applied to two binomials.

### Special products

$$
\begin{aligned}
(a + b)^2 &= a^2 + 2ab + b^2 \\
(a - b)^2 &= a^2 - 2ab + b^2 \\
(a + b)(a - b) &= a^2 - b^2
\end{aligned}
$$

```python
fig, ax = plt.subplots(figsize=(5.2, 5.2))
a, b = 3.0, 1.6
cells = [
    (0, b, a, a, BLUE, r"$a^2$"),
    (a, b, b, a, ORANGE, r"$ab$"),
    (0, 0, a, b, ORANGE, r"$ab$"),
    (a, 0, b, b, GREEN, r"$b^2$"),
]
for x, y, w, h, c, lab in cells:
    ax.add_patch(Rectangle((x, y), w, h, facecolor=c + "30",
                           edgecolor=c, lw=1.8))
    ax.text(x + w / 2, y + h / 2, lab, ha="center", va="center",
            fontsize=18, color=INK)
# side labels
ax.text(a / 2, a + b + 0.15, "$a$", ha="center", fontsize=14)
ax.text(a + b / 2, a + b + 0.15, "$b$", ha="center", fontsize=14)
ax.text(-0.2, b + a / 2, "$a$", ha="right", va="center", fontsize=14)
ax.text(-0.2, b / 2, "$b$", ha="right", va="center", fontsize=14)
ax.set_xlim(-0.6, a + b + 0.2)
ax.set_ylim(-0.9, a + b + 0.6)
ax.set_aspect("equal")
ax.axis("off")
ax.text((a + b) / 2, -0.55,
        r"$(a+b)^2 = a^2 + 2ab + b^2$   (the two $ab$ pieces are the middle term)",
        ha="center", fontsize=10.5, color=INK)
plt.show()
```

*Area model for $(a + b)^2$. Dropping the middle term means losing both $ab$ rectangles, which is why $(a + b)^2 \neq a^2 + b^2$.*

**Worked example.**

$$
(x - \mu)^2 = x^2 - 2\mu x + \mu^2
$$

:::{admonition} Common mistakes
:class: warning
- Forgetting the middle term: $(x + 3)^2 = x^2 + 6x + 9$, not $x^2 + 9$.
- Sign errors when subtracting: $(3x^2 + x) - (x^2 - 4x) = 2x^2 + 5x$.
- Missing a term when multiplying longer polynomials. A grid (one polynomial across the top, one down the side) prevents this.
:::

:::{tip} Where this shows up
Expanding $(x - \mu)^2$ is the algebra behind the variance identity $\operatorname{Var}(X) = E[X^2] - \mu^2$ and behind least-squares derivations.
:::

## 1.5 Factoring Polynomials

Factoring reverses multiplication. Check your answer by multiplying back out.

### Decision order

1. **Pull out the GCF first, always.** Take the GCF of the coefficients and the lowest power of each shared variable.
2. Count the terms:
   - **Two terms:** difference of squares, or sum/difference of cubes.
   - **Three terms:** perfect square trinomial, or a general trinomial.
3. Check whether any factor can be factored further.

### Trinomials

- **Leading coefficient 1,** $x^2 + bx + c$: find $p, q$ with $pq = c$ and $p + q = b$. Then $x^2 + bx + c = (x + p)(x + q)$.
- **General,** $ax^2 + bx + c$ (grouping, or "$ac$ method"): find $p, q$ with $pq = ac$ and $p + q = b$. Split the middle term into $px + qx$, then factor in pairs.

**Worked example.** Factor $2x^2 + 5x + 3$. Here $ac = 6$, and $2 \cdot 3 = 6$ with $2 + 3 = 5$.

$$
2x^2 + 2x + 3x + 3 = 2x(x + 1) + 3(x + 1) = (2x + 3)(x + 1)
$$

```python
fig, (a1, a2) = plt.subplots(1, 2, figsize=(9.5, 3.9))

x = np.linspace(-3, 1, 400)
a1.plot(x, 2 * x ** 2 + 5 * x + 3, color=BLUE, lw=2.4)
for r, lab, dx in [(-1.5, r"$2x+3=0$" + "\n" + r"$x=-\frac{3}{2}$", -0.9),
                   (-1.0, r"$x+1=0$" + "\n" + r"$x=-1$", 0.45)]:
    a1.plot(r, 0, "o", color=ORANGE, ms=8, zorder=3)
    a1.annotate(lab, (r, 0), xytext=(r + dx, -3.2), color=ORANGE, fontsize=10,
                arrowprops=dict(arrowstyle="->", color=ORANGE))
center_axes(a1)
a1.set_ylim(-3.6, 6)
a1.set_title(r"$2x^2+5x+3 = (2x+3)(x+1)$", color=INK)

x = np.linspace(-2.7, 2.7, 400)
a2.plot(x, x ** 4 - 16, color=BLUE, lw=2.4)
for r in (-2, 2):
    a2.plot(r, 0, "o", color=ORANGE, ms=8, zorder=3)
a2.text(-1.45, 14, r"$x = \pm 2$ from $(x+2)(x-2)$", color=ORANGE, fontsize=10)
a2.text(-1.2, -9.5, r"$x^2 + 4$ is never $0$:" + "\nno more intercepts", color=GRAY, fontsize=10)
a2.spines["bottom"].set_position("zero")
a2.set_ylim(-20, 30)
a2.set_title(r"$x^4-16 = (x^2+4)(x+2)(x-2)$", color=INK)
fig.tight_layout()
plt.show()
```

*Each linear factor is an $x$-intercept: setting $(2x + 3)(x + 1) = 0$ gives $x = -\tfrac{3}{2}$ and $x = -1$. On the right, $x^2 + 4$ adds no intercepts because a sum of squares is never zero. Factoring and finding where a graph crosses zero are the same problem.*

### Special forms

$$
\begin{aligned}
a^2 + 2ab + b^2 &= (a + b)^2 \\
a^2 - b^2 &= (a + b)(a - b) \\
a^3 + b^3 &= (a + b)(a^2 - ab + b^2) \\
a^3 - b^3 &= (a - b)(a^2 + ab + b^2)
\end{aligned}
$$

- Cube signs follow **SOAP**: **S**ame, **O**pposite, **A**lways **P**ositive.
- A **sum of squares** $a^2 + b^2$ does not factor over the reals.

### Fractional or negative exponents

Factor out the common base raised to its **lowest** exponent, then subtract that exponent from each term's exponent.

**Worked example.**

$$
3x(x + 2)^{-1/2} + 2(x + 2)^{1/2} = (x + 2)^{-1/2}\big[3x + 2(x + 2)\big] = (x + 2)^{-1/2}(5x + 4)
$$

:::{admonition} Common mistakes
:class: warning
- Skipping the GCF step, which makes everything else harder.
- Stopping early: $x^4 - 16 = (x^2 + 4)(x^2 - 4) = (x^2 + 4)(x + 2)(x - 2)$.
- With fractional exponents, choosing the larger exponent as the GCF. The lowest is always the right one, even when it is negative.
:::

:::{tip} Where this shows up
Derivatives from the product and chain rules come out looking exactly like the fractional-exponent example, and factoring them is how you find where they equal zero. Factoring also resolves many $0/0$ limits in Stewart.
:::

## 1.6 Rational Expressions

A **rational expression** is a quotient of polynomials, $\dfrac{P(x)}{Q(x)}$ with $Q(x) \neq 0$. Every operation works like ordinary fractions, and the key step is almost always **factor first**.

| Operation | Method |
|---|---|
| Simplify | Factor the top and bottom, then cancel common **factors** |
| Multiply | Factor, cancel, multiply across |
| Divide | Multiply by the reciprocal of the second expression |
| Add / subtract | Rewrite over the LCD, then combine numerators |
| Complex fraction | Multiply the top and bottom by the LCD of every inner fraction |

- The **LCD** is the product of every distinct factor, each at the highest power it appears.
- Values that make any original denominator zero stay excluded, even after cancelling.

**Worked example (subtraction).**

$$
\begin{aligned}
\frac{3}{x - 1} - \frac{x + 1}{x^2 - 1}
&= \frac{3(x + 1)}{(x - 1)(x + 1)} - \frac{x + 1}{(x - 1)(x + 1)} \\
&= \frac{3x + 3 - x - 1}{(x - 1)(x + 1)}
= \frac{2(x + 1)}{(x - 1)(x + 1)}
= \frac{2}{x - 1}, \quad x \neq \pm 1
\end{aligned}
$$

**Worked example (complex fraction).** Multiply the top and bottom by $xy$:

$$
\frac{\frac{1}{x} - \frac{1}{y}}{\frac{1}{x} + \frac{1}{y}} = \frac{y - x}{y + x}
$$

```python
fig, ax = plt.subplots(figsize=(7.5, 4.4))
for lo, hi in [(-4, 0.92), (1.08, 5)]:
    x = np.linspace(lo, hi, 400)
    ax.plot(x, 2 / (x - 1), color=BLUE, lw=2.4)
ax.axvline(1, color=GRAY, ls="--", lw=1.2)
ax.plot(-1, -1, "o", ms=9, mfc="white", mec=BLUE, mew=2, zorder=4)
ax.annotate("hole at $(-1, -1)$:\nfactor $x+1$ cancelled,\nbut $x=-1$ is still excluded",
            (-1, -1), xytext=(-3.9, 3.2), fontsize=10, color=INK,
            arrowprops=dict(arrowstyle="->", color=INK))
ax.text(1.6, 5.6, "vertical asymptote $x = 1$:\n$x-1$ stays in the denominator",
        fontsize=10, color=GRAY)
center_axes(ax)
ax.set_ylim(-9, 9)
ax.set_title(r"$\frac{3}{x-1} - \frac{x+1}{x^2-1} \;=\; \frac{2}{x-1},\quad x \neq \pm 1$",
             color=INK, fontsize=13)
plt.show()
```

*The simplified $\dfrac{2}{x - 1}$ graphs the same as the original *except* at $x = -1$, where there is a hole. A factor that cancels leaves a hole; a factor left in the denominator leaves a vertical asymptote. (Asymptotes come back in Chapter 5.)*

:::{admonition} Common mistakes
:class: warning
- Cancelling **terms** instead of **factors**: $\dfrac{x + 2}{x}$ does not simplify to $2$. You can only cancel something that multiplies the *entire* numerator and the *entire* denominator.
- When subtracting, the minus applies to the whole second numerator. Put it in parentheses before distributing.
- Adding straight across: $\dfrac{1}{a} + \dfrac{1}{b} \neq \dfrac{1}{a + b}$. The correct sum is $\dfrac{a + b}{ab}$.
:::

:::{tip} Where this shows up
Limits in Stewart (factor, cancel, then substitute), partial fractions for integration, and the algebra of Bayes' rule, which is a ratio you simplify by cancelling shared factors.
:::
