```python
import yaml, os, glob, subprocess

ROOT = "/Users/colelacroix/math"

# The ONLY thing you maintain by hand: folder name -> book title.
# Add a line here when you start a new subject. That's it.
BOOKS = {
    "0.1_think_bayes_(Downey)": "Think Bayes (Downey)",
    "1. Algebra and Trigonometry": "Algebra and Trigonometry",
}

def rebuild_toc():
    """Scans BOOKS folders for notebooks and rewrites myst.yml correctly."""
    toc = [{"file": "index.md"}]
    for folder, title in BOOKS.items():
        folder_path = os.path.join(ROOT, folder)
        if not os.path.isdir(folder_path):
            continue
        notebooks = sorted(glob.glob(os.path.join(folder_path, "*.ipynb")))
        if not notebooks:
            continue
        children = [{"file": os.path.relpath(nb, ROOT)} for nb in notebooks]
        toc.append({"title": title, "children": children})

    config = {
        "version": 1,
        "project": {
            "id": "af601852-cbc3-4d41-9731-bb9a87aa495e",
            "title": "Math Notes",
            "description": "Notes and worked examples",
            "github": "https://github.com/coletlacroix/math",
            "toc": toc,
        },
        "site": {"template": "book-theme"},
    }

    with open(os.path.join(ROOT, "myst.yml"), "w") as f:
        yaml.dump(config, f, sort_keys=False, allow_unicode=True)
    print("myst.yml rebuilt from current folders")

def push(message):
    """Rebuilds the toc, then commits and pushes everything."""
    rebuild_toc()
    subprocess.run(["git", "-C", ROOT, "add", "."])
    subprocess.run(["git", "-C", ROOT, "commit", "-m", message])
    subprocess.run(["git", "-C", ROOT, "push", "origin", "main"])
```
