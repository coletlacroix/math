# Bayes's Theorem

#### The Cookie Problem

- Bowl 1: 30 vanilla, 10 chocolate cookies
- Bowl 2: 20 vanilla, 20 chocolate cookies

If we choose, at random, a bowl and a cookie from that bowl, drawing vanilla, what is the probability that it came from bowl 1?

We are looking for the conditional probability:

$$ P(B_1 \mid V) $$

However, what we get from the statement is:

$$ P(V \mid B_1) $$

and

$$ P(V \mid B_2) $$

Bayes's theorum tells us how they are related:

$$ P(B_1 \mid V) = \frac {P(B_1) \, P(V\mid B_1)} {P(V)} $$

So, in order to complete the equation, we need:

$$ P(V) $$

Which we can find, using the law of total probability:

$$ P(V) = P(B_1) \, P(V\mid B_1) \; + \; P(B_2) \, P(V \mid B_2) $$

Using what we know, we can fill it in:

$$ P(V) = (1/2) \, (3/4) \; + \; (1/2) \, (1/2) = (5/8) $$

So, using that new information, we can derive $$ P(B_1 \mid V) $$ by plugging in the numbers:

$$ P(B_1 \mid V) = \frac{(1/2)\, (3/4)}{(5/8)} = 3/5 $$

#### Diachronic Bayes (related to change over time)

H = hypothesis (prior)
D = Body of data

$$ P(H \mid D) = \frac {P(H) \, P(D\mid H)}{ P(D)} $$

Portions of the equation:

Prior is the probability of the data before we see the hypothesis: $$ P(H) $$

Posterior is the probability of the hypothesis after we see the data: $$ P(H \mid D) $$

Likelihood is the probability of the data under the hypothesis: $$ P(D \mid H) $$

Total probability of the data is: $$ P(D) $$

As in the previous example, we can compute $$ P(D) $$

using the law of total probability. For example with two hypotheses:

$$ P(D) = P(H_1) \, P(D \mid H_1) + P(H_2) \, P(D \mid H_2) $$

or with any number of hypotheses:

$$ P(D) = \sum_ {i} P(H_i) \; P(D \mid H_i) $$

#### Bayes Tables

```python
import pandas as pd

table = pd.DataFrame(index=['Bowl 1', 'Bowl 2'])

table['prior'] = 1/2, 1/2
table
```

```python
table['likelihood'] = 3/4, 1/2
table
```

```python
table['unnorm'] = table['prior'] * table['likelihood']
table
```

Calling them 'unnorm' because they are unnormalized posteriors, the product of a prior and a likelihood 

$$ P(B_i) \, P(D \mid B_i) $$

which is the numberator of Bayes's theorem. 

If you add them up, 

$$ P(B_1) \, P(D \mid B_1) + P(B_2) \, P(D \mid B_2) $$

you have the denominator, 

$$ P(D) $$ 

```python
# Let's compute total probability 

prob_data = table['unnorm'].sum()
prob_data
```

```python
# and posterior probabilities

table['posterior'] = table['unnorm'] / prob_data
table
```

Here we have found the posterior probabilities of both bowls having drawn a vanilla!

#### The Dice Problem

A Bayes table can also solve problems with more than two hypotheses. Consider this example:

- I have a box with a 6-sided, 8-sided, and 12-sided die. If you choose one of the die at random, roll it, and find that you have rolled a 1, what is the probability that you have chose the 6-sided die?

In this example, you have three hypotheses with equal prior probabilities (just as likely that you drew from the box the 6, 8, or 12-sided die). 

However, the data (rolling a "1") gives us new insights. 

```python
table2 = pd.DataFrame(index=[6,8,12])
```

```python
from fractions import Fraction

table2['prior'] = Fraction(1,3)
table2['likelihood'] = Fraction(1,6), Fraction(1,8), Fraction(1,12)
table2
```

```python
def update(table):
    table['unnorm'] = table ['prior'] * table ['likelihood']
    prob_data = table ['unnorm'].sum()
    table['posterior'] = table ['unnorm'] / prob_data
    return prob_data

prob_data = update(table2)

table2
```

We can see here with our posteriors that the 6-sided die is most likely because drawing a 1 on that die has highest likelihood in our table. 

#### The Monty Hall Problem

Consider this example:

- You see three numbered doors with a prize behind one of them.
- The objective of the game is to choose the door with the prize.
- After selecting a door, one of the remaining doors is opened to show no prize behind it.
- You are given the option to switch your selected door.

To maximize your chance of winning the prize, do you stick with your selected door or choose the remaining door?

```python
table3 = pd.DataFrame(index=['Door 1', 'Door 2', 'Door 3'])
table3['prior'] = Fraction(1,3)
table3
```

```python
table3['likelihood'] = Fraction(1,2), 1, 0
table3
```

```python
# now let's use the update function
update(table3)
table3
```
